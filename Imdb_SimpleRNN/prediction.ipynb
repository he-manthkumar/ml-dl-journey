```python
word_index = imdb.get_word_index()
og={value:key for key,value in word_index.items()}
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Embedding,SimpleRNN
```

```python
from tensorflow.keras.models import load_model
```

```python
model= load_model('imdb_rnn.h5')
```

```python
model.summary()
```

```python
model.get_weights()
```

```python
def decode_review(encoded_review):
    decoded_review = ' '.join([og.get(i-3,'?') for i in encoded_review])
    return decoded_review
def preprocess_text(text):
    words = text.lower().split()    
    encoded_review = [word_index.get(word,2)+3 for word in words]
    padded_review = sequence.pad_sequences([encoded_review],maxlen=500)
    return padded_review
```

```python
def predict_sentiment(text):
    encoded_text = preprocess_text(text)
    prediction = model.predict(encoded_text)
    sentiment = 'Good' if prediction[0][0]>0.5 else 'Bad'
    return sentiment,prediction[0][0]
```

```python
example_text = "The movie was a great waste of my time but i loved it"
print(predict_sentiment(example_text))
```
